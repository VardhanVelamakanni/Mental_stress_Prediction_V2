```python
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
```

```python
data_dir = "/content"

train_files = np.load(f"{data_dir}/train_files.npy", allow_pickle=True)
val_files   = np.load(f"{data_dir}/val_files.npy", allow_pickle=True)
test_files  = np.load(f"{data_dir}/test_files.npy", allow_pickle=True)

y_train = np.load(f"{data_dir}/y_train.npy")
y_val   = np.load(f"{data_dir}/y_val.npy")
y_test  = np.load(f"{data_dir}/y_test.npy")

print("Train size:", len(train_files))
print("Val size:", len(val_files))
print("Test size:", len(test_files))
```

```python
class EEGGridDataset(Dataset):
    def __init__(self, files, labels):
        self.files = files
        self.labels = labels

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        x = np.load(self.files[idx])
        x = torch.tensor(x).permute(2,0,1).float()  # (5,32,32)
        y = torch.tensor(self.labels[idx]).long()
        return x, y
```

```python
train_ds = EEGGridDataset(train_files, y_train)
val_ds   = EEGGridDataset(val_files, y_val)
test_ds  = EEGGridDataset(test_files, y_test)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)
val_loader   = DataLoader(val_ds, batch_size=64)
test_loader  = DataLoader(test_ds, batch_size=64)
```

```python
class ResidualBlock(nn.Module):
    def __init__(self, channels):
        super().__init__()
        self.conv1 = nn.Conv2d(channels, channels, 3, padding=1)
        self.bn1 = nn.BatchNorm2d(channels)
        self.conv2 = nn.Conv2d(channels, channels, 3, padding=1)
        self.bn2 = nn.BatchNorm2d(channels)

    def forward(self, x):
        out = torch.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        return torch.relu(out + x)


class EEGNetPlus(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer1 = nn.Sequential(
            nn.Conv2d(5, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.res1 = ResidualBlock(32)

        self.layer2 = nn.Sequential(
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        self.res2 = ResidualBlock(64)

        self.layer3 = nn.Sequential(
            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d(1)
        )

        self.fc = nn.Linear(128, 2)

    def forward(self, x):
        x = self.layer1(x)
        x = self.res1(x)
        x = self.layer2(x)
        x = self.res2(x)
        x = self.layer3(x)
        x = x.view(x.size(0), -1)
        return self.fc(x)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = EEGNetPlus().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
```

```python
best_val = 0

for epoch in range(30):

    model.train()
    train_correct = 0
    train_total = 0

    for x,y in train_loader:
        x,y = x.to(device), y.to(device)

        optimizer.zero_grad()
        out = model(x)
        loss = criterion(out,y)
        loss.backward()
        optimizer.step()

        train_correct += (out.argmax(1)==y).sum().item()
        train_total += y.size(0)

    train_acc = train_correct/train_total

    model.eval()
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for x,y in val_loader:
            x,y = x.to(device), y.to(device)
            out = model(x)
            val_correct += (out.argmax(1)==y).sum().item()
            val_total += y.size(0)

    val_acc = val_correct/val_total

    print(f"Epoch {epoch+1} | Train Acc: {train_acc:.4f} | Val Acc: {val_acc:.4f}")

    if val_acc > best_val:
        best_val = val_acc
        torch.save(model.state_dict(),"best_model.pth")
```

```python
model.load_state_dict(torch.load("best_model.pth"))
model.eval()

correct = 0
total = 0

with torch.no_grad():
    for x,y in test_loader:
        x,y = x.to(device), y.to(device)
        out = model(x)
        correct += (out.argmax(1)==y).sum().item()
        total += y.size(0)

print("FINAL TEST ACCURACY:", correct/total)
```

```python
import torch
import numpy as np

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = EEGNetPlus().to(device)
model.load_state_dict(torch.load("best_model.pth"))
model.eval()
```

```python
def predict(sample_path):
    x = np.load(sample_path)
    x = torch.tensor(x).permute(2,0,1).unsqueeze(0).float().to(device)

    with torch.no_grad():
        out = model(x)
        pred = out.argmax(1).item()

    if pred == 0:
        print("Prediction: LOW / NORMAL MENTAL STRESS")
    else:
        print("Prediction: HIGH MENTAL STRESS")
```
